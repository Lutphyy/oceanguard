````python
# 🚀 CELL 4: Train YOLO dengan OPTIMAL SETTINGS
# Waktu: ~30-90 menit (tergantung dataset & GPU)

from ultralytics import YOLO
import torch
import numpy as np

# Verify numpy version
print(f"🔍 Numpy version: {np.__version__}")
if np.__version__ != '1.26.4':
    print("\n❌ ERROR: Wrong numpy version!")
    print("⚠️  You MUST restart runtime after Cell 1!")
    print("⚠️  Steps:")
    print("   1. Runtime → Restart runtime")
    print("   2. Re-run Cell 1")
    print("   3. Re-run Cell 1B to verify")
    print("   4. Then run Cell 2, 3, 4")
    raise RuntimeError("Wrong numpy version. Restart runtime required.")

print("✅ Numpy version OK\n")
print("🚀 Starting ULTIMATE Training...\n")
print("📊 Training Configuration:")
print("  - Model: YOLOv8 Nano (stable & reliable)")
print("  - Pretrained: MS COCO")
print("  - Epochs: 200 (with early stopping)")
print("  - Patience: 20 (auto-stop if no improvement)")
print("  - Augmentation: STRONG (for better generalization)")
print("  - Optimizer: AdamW (best for YOLO)")
print("\n⏱️  Estimated time: 30-90 minutes")
print("☕ Grab a coffee and relax!\n")
print("="*60)

# Load pretrained YOLOv8 Nano with MS COCO weights
# YOLOv8 has stable auto-download (YOLOv11 has download issues)
model = YOLO('yolov8n.pt')  # Will auto-download (~6 MB)

# ULTIMATE Training with OPTIMAL SETTINGS
results = model.train(
    # Dataset
    data=f'{dataset.location}/data.yaml',
    
    # Training Duration
    epochs=200,              # More epochs for better convergence
    patience=20,             # Early stopping (prevent overfitting)
    
    # Batch & Image Size
    batch=16,                # Optimal for T4 GPU
    imgsz=640,               # Standard YOLO size
    
    # Device
    device=0,                # Use GPU
    
    # Optimizer
    optimizer='AdamW',       # Best optimizer for YOLO
    lr0=0.001,               # Initial learning rate
    lrf=0.01,                # Final learning rate (1% of lr0)
    momentum=0.937,          # SGD momentum
    weight_decay=0.0005,     # L2 regularization
    warmup_epochs=3.0,       # Warmup for stable training
    warmup_momentum=0.8,
    warmup_bias_lr=0.1,
    
    # Loss Weights
    box=7.5,                 # Box loss weight
    cls=0.5,                 # Class loss weight
    dfl=1.5,                 # Distribution focal loss weight
    
    # Augmentation (STRONG for better generalization)
    hsv_h=0.015,             # Hue augmentation
    hsv_s=0.7,               # Saturation augmentation
    hsv_v=0.4,               # Value (brightness) augmentation
    degrees=15.0,            # Rotation ±15°
    translate=0.1,           # Translation 10%
    scale=0.5,               # Scale 50%
    shear=0.0,               # No shear (underwater images)
    perspective=0.0,         # No perspective (underwater)
    flipud=0.0,              # No vertical flip (underwater orientation)
    fliplr=0.5,              # Horizontal flip 50%
    mosaic=1.0,              # Mosaic augmentation (IMPORTANT!)
    mixup=0.1,               # Mixup augmentation 10%
    copy_paste=0.0,          # No copy-paste
    
    # Training Settings
    pretrained=True,         # Use pretrained weights
    verbose=True,            # Show training progress
    plots=True,              # Generate plots
    save=True,               # Save checkpoints
    save_period=10,          # Save every 10 epochs
    amp=True,                # Automatic Mixed Precision (faster)
    
    # Project Settings
    project='ocean_trash',
    name='ultimate_yolov8n',
    exist_ok=True,
    
    # Advanced
    close_mosaic=10,         # Disable mosaic in last 10 epochs
    label_smoothing=0.0,     # No label smoothing
    nbs=64,                  # Nominal batch size
    cos_lr=False,            # No cosine LR scheduler
    fraction=1.0,            # Use 100% of dataset
    profile=False,           # No profiling
    freeze=None,             # No frozen layers
    resume=False,            # Start fresh
    seed=42,                 # Random seed for reproducibility
    deterministic=True,      # Deterministic training
)

print("\n" + "="*60)
print("✅ Training completed successfully!")
print(f"📁 Best weights: {model.trainer.best}")
print(f"📁 Last weights: {model.trainer.last}")
print("\n🎉 Ready for evaluation!")
````

````python
# 📊 CELL 5: Evaluate Model Performance
# Waktu: ~1-2 menit

import os
from ultralytics import YOLO

print("🔍 Evaluating model on validation set...\n")

# Auto-detect correct path (yolov8n or yolov11n)
if os.path.exists('ocean_trash/ultimate_yolov8n/weights/best.pt'):
    model_path = 'ocean_trash/ultimate_yolov8n/weights/best.pt'
    project_name = 'ultimate_yolov8n'
elif os.path.exists('ocean_trash/ultimate_yolov11n/weights/best.pt'):
    model_path = 'ocean_trash/ultimate_yolov11n/weights/best.pt'
    project_name = 'ultimate_yolov11n'
else:
    print("❌ ERROR: Model weights not found!")
    print("\n📁 Available folders:")
    !ls -la ocean_trash/
    raise FileNotFoundError("Model weights not found. Did Cell 4 complete successfully?")

print(f"✅ Found model: {model_path}\n")

# Load best model
best_model = YOLO(model_path)

# Validate
metrics = best_model.val()

# Calculate F1 Score
precision = metrics.box.mp
recall = metrics.box.mr
f1_score = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0

print("\n" + "="*60)
print("📈 VALIDATION METRICS:")
print("="*60)
print(f"mAP@50:     {metrics.box.map50:.1%}  {'✅' if metrics.box.map50 > 0.80 else '⚠️'}")
print(f"mAP@50-95:  {metrics.box.map:.1%}")
print(f"Precision:  {precision:.1%}  {'✅' if precision > 0.80 else '⚠️'}")
print(f"Recall:     {recall:.1%}  {'✅' if recall > 0.70 else '⚠️'}")
print(f"F1 Score:   {f1_score:.1%}  {'✅' if f1_score > 0.75 else '⚠️'}")
print("="*60)

# Compare with Roboflow
print("\n📊 COMPARISON WITH ROBOFLOW:")
print("="*60)
roboflow_map = 0.816
roboflow_prec = 0.846
roboflow_rec = 0.743
roboflow_f1 = 0.791

map_diff = metrics.box.map50 - roboflow_map
prec_diff = precision - roboflow_prec
rec_diff = recall - roboflow_rec
f1_diff = f1_score - roboflow_f1

print(f"mAP@50:    {map_diff:+.1%}  {'🎉 BETTER!' if map_diff > 0 else '⚠️ Lower'}")
print(f"Precision: {prec_diff:+.1%}  {'🎉 BETTER!' if prec_diff > 0 else '⚠️ Lower'}")
print(f"Recall:    {rec_diff:+.1%}  {'🎉 BETTER!' if rec_diff > 0 else '⚠️ Lower'}")
print(f"F1 Score:  {f1_diff:+.1%}  {'🎉 BETTER!' if f1_diff > 0 else '⚠️ Lower'}")
print("="*60)

# Per-class metrics
print("\n📊 PER-CLASS METRICS:")
print("="*60)
for i, class_name in enumerate(metrics.names.values()):
    class_metrics = metrics.box.class_result(i) if hasattr(metrics.box, 'class_result') else None
    if class_metrics:
        print(f"{class_name:12s}: P={class_metrics[0]:.1%} R={class_metrics[1]:.1%} mAP@50={class_metrics[2]:.1%}")
print("="*60)

print("\n✅ Evaluation complete!")
print(f"\n📁 Training plots saved to: ocean_trash/{project_name}/")
````

## 📊 Metrics Guide

- **mAP@50**: Mean Average Precision @ IoU 0.5
  - > 80%: Excellent ⭐⭐⭐⭐⭐
  - 70-80%: Good ⭐⭐⭐⭐
  - 60-70%: Fair ⭐⭐⭐

- **Precision**: Accuracy of detections
  - > 80%: Excellent (few false positives)
  - 70-80%: Good
  - < 70%: Needs improvement

- **Recall**: Coverage of objects
  - > 75%: Excellent (few missed objects)
  - 65-75%: Good
  - < 65%: Needs improvement
